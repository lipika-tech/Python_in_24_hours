```python
# recurssion
def fact(i):
    return 1 if(i==1 or i==0) else i*fact(i-1)
```

```python
for num in range(5,11):
    print("factorial of",num,"is :", fact(num))
```

```python
def factorial(num):
    if (num == 1 or num == 0):
        return 1
    else:
        return (num * factorial(num - 1)) 
```

```python
#Driver Code 
num = 7 
print("number: ",num)
print("Factorial: ",factorial(num))
```
