## Matrix

```python
mymat1 = np.asmatrix('3, 2, 1;4, 1, 5;4, 1, 7')
```

```python
print(mymat1)
```

```python
type(mymat1)
```

```python
# Transpose
mymat1.T
```

```python
mymat_inv = np.linalg.inv(mymat1)
```

```python
print(mymat_inv)
```

```python
# Determinant of a matrix
mymat_det = np.linalg.det(mymat1)
```

```python
print(mymat_det)
```
