```python
# Concept of OOPS:

# Class = a user defined data type which holds real time objects --- e.g. Mobile
# mobile is similar to pre defined data type like int a, string 'jalal', etc.

# Methods (Functions) = it stores tasks that object will perform (behaviour) --- e.g. making calls, playing games, watching video. etc.

# Objects (Instances of Class) = a class can have multiple objects --- e.g. Apple, Samsung, RealMe, Xiaomi, etc.
# Apple is similar to value of a predefined data type like a = 10
```

```python
class Classroom:
    def mark_attendance(self):
        print("I have marked the attendance")
    
    def study_well(self):
        print("I am going to study well for 8 hours a day")
        
    def sleep_well(self):
        print(" I am going to sleep well for focused attention")
```

```python
C1 = Classroom()
```

```python
C1.mark_attendance()
```

```python
C1.study_well()
```

```python
C1.sleep_well()
```
