```python
class Animal:
    name=""
    def eat(self):
        print("yes i do eat")







class Dog(Animal):
    def eat(self):
        super().eat()
        print("i like it")
        
    def display(self):
        print("name display",self.name)



dog1=Dog()

# acsess from superclass

dog1.name="Puppy"
dog1.eat() #method overriding

# acess of subclass
dog1.display()




```
