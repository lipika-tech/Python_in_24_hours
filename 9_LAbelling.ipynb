```python
#Adding title and Labelling the Axes in the graph

# importing matplotlib module
from matplotlib import pyplot as plt

# x-axis values
x = [5, 2, 9, 4, 7]

# Y-axis values
y = [10, 5, 8, 4, 2]

# Function to plot
plt.bar(x, y)

# Adding Title
plt.title("Demo Class Graph")

# Labeling the axes
plt.xlabel("Time")
plt.ylabel("Distance")

# function to show the plot
plt.show()
```
