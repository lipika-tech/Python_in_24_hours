```python
# importing the required module
import matplotlib.pyplot as plt
# x axis values
x = [1,2,3]
# corresponding y axis values
y = [2,4,1]
# plotting the points
plt.plot(x, y)
# naming the x axis
plt.xlabel('Hello World')
# naming the y axis
plt.ylabel('Python')
# giving a title to my graph
plt.title('Day 5 demo session')
# function to show the plot
plt.show()
```
