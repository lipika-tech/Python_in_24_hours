```python
# Ticks are the markers denoting data points on axes.
# importing libraries
import matplotlib.pyplot as plt
import numpy as np

# values of x and y axes
x = [5, 10, 15, 20, 25, 30, 35, 40, 45, 50]
y = [1, 4, 3, 2, 7, 6, 9, 8, 10, 5]

plt.plot(x, y, 'g')
plt.xlabel('Dist')
plt.ylabel('speed')
# here we set the size for ticks, rotation and color value

plt.tick_params(axis="x", labelsize=20, labelrotation=30, labelcolor="red")
plt.tick_params(axis="y", labelsize=20, labelrotation=45, labelcolor="green")

plt.show()
```
