```python
#Line plot using matplotlib librabry, here we use pyplot function of matplotlib
# importing matplotlib module
from matplotlib import pyplot as plt

# x-axis values
x = [5, 2, 6, 4, 7]

# Y-axis values
y = [10, 9, 8, 4, 2]

# Function to plot
plt.scatter(x, y)

# function to show the plot
plt.show()
```
