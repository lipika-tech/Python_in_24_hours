```python
"""
# multiple inheritance

class Animal:
    def eat(self):
        print(" i eat")


class Mammal:
    def info(self):
        print("I am Mammal")


class Dog(Animal, Mammal):
    def display(self):
        print("hello i'm dog")



dog1=Dog()
dog1.display() # dog class
dog1.eat()# Animal
dog1.info() #Mammal



#MRO

class Animal:
    def eat(self):
        print(" Animal class: eat")


class Mammal:
    def eat(self):
        print("Mammal class : eat")
    def info(self):
        print("I am Mammal")


class Dog(Mammal, Animal):
    def display(self):
        print("hello i'm dog")

dog1=Dog()
dog1.display() # dog class
dog1.eat()# Animal
dog1.info() #Mammal


"""
# multilevel Inheritance

class superclass:
    def super_method(self):
        print("super class")

class derivedclass1(superclass):
    def derived_method1(self):
        print("derived method 1 class")

    
class derivedclass2(derivedclass1):
    def derived_method2(self):
        print("derived method 2 class")


obj1=derivedclass2()
obj1.super_method()# superclass
obj1.derived_method1()# derived class 1
obj1.derived_method2() # derived class 2

```
