## List

```python
# initial list
scores = [80,90,95,100]

# memory location of initial list
print('Memory location initially is ' , id(scores))
print('='*20)

```

```python
scores = [80,90,95,100]
# change value
scores[2] = 99
# display modified list
print(scores)
print('='*20)
# check memory location of modified list 
print('Memory location of modified list is ' , id(scores))
```

```python
A={0,2,4,6,8,7}

B={1,2,3,4,5}

print('union',A|B)  # 0,1,2,3,4,5,6,7,8
print('intersection',A&B) #2,4 
print('set difference A-B', A-B) #0,6,8,7
print('set difference B-A', B-A)#1,3,5
print('symmetric differnce',A^B) # 0,1,3,5,6,7,8
```
