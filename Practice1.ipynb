```python
def fact(i):
    return 1 if(i==1 or i==0) else i*fact(i-1)
```

```python
num=5
print(fact(num))
```

```python
#between 5 to 10
for i in range(5,11):
    print(i)
```

```python
for num in range(5,11):
    print("factorial of",num,"is :", fact(num))
```
