```python
import matplotlib.pyplot as plt
import numpy as np
x = np.arange(0, 10, 0.1)
y = np.sin(x)

# figures in matplotlib using figure()
plt.figure(figsize=(5, 2))
plt.grid()
plt.plot(x, y)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
x = np.arange(0, 10, 0.1)
y = np.cos(x)

# figures in matplotlib using figure()
plt.figure(figsize=(3, 3))
plt.plot(x, y)
plt.show()
```

```python
x=np.linspace(0,30,200)
y1=np.sin(x)
y2=np.cos(x)
fig,ax=plt.subplots()
ax.set_xscale("log")
ax.plot(x,y2)
plt.show()
```

```python
# Pair Plots is seaborn - distribution of curve
x=np.linspace(0,30,2000)
y1=np.sin(x)
y2=np.cos(x)
fig,ax=plt.subplots()
ax.plot(x,y1)
ax.plot(x,y2)
plt.show()
```

```python
x=np.linspace(0,30,2000)
y1=np.sin(x)
y2=np.cos(x)
fig,ax=plt.subplots()
ax.plot(x[:1000],y1[:1000])
ax.plot(x[1000:],y2[1000:])
ax.fill_betweenx([0,1],[0],[4])
plt.show()
```
