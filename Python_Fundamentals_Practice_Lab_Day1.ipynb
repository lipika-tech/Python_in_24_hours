- `#` starts a single-line comment. Triple quotes `""" ... """` are used for docstrings / multi-line notes.
- **Indentation is syntax.** Python uses indentation (4 spaces recommended) instead of `{ }` to define blocks. Wrong indentation → `IndentationError`.


````python
# 🎯 Demo 2.1 — Comments and indentation

# This is a single-line comment: the interpreter ignores it
"""
This is a multi-line string often used as a block note or docstring.
"""

billable_hours = 42   # inline comment

if billable_hours > 40:
    print("Overtime recorded")        # indented → inside the if-block
    print("Notify project manager")   # same block
print("This line always runs")        # back to no indentation → outside the block

# Uncomment the next two lines to see an IndentationError:
# if True:
# print("not indented")
````

### 2.2 Variables and naming conventions
A variable is a **name that refers to a value**. Create it simply by assigning: `name = value`.

| Rule (must) | Convention – PEP 8 (should) |
|---|---|
| Letters, digits, `_` only | `snake_case` for variables & functions → `employee_name` |
| Cannot start with a digit | `UPPER_CASE` for constants → `MAX_HOURS = 40` |
| Case-sensitive (`Name` ≠ `name`) | `PascalCase` for classes → `EmployeeRecord` |
| Cannot be a keyword (`if`, `for`, `class`…) | Descriptive names: `hourly_rate`, not `x` |


````python
# 🎯 Demo 2.2 — Valid names, multiple assignment, swapping
employee_name = "Priya Sharma"
employee_id = 10234
MAX_WEEKLY_HOURS = 40        # constant by convention

# Multiple assignment in one line
city, country = "Chicago", "USA"

# Same value to many variables
q1 = q2 = q3 = 0

# Swap without a temporary variable
a, b = 5, 10
a, b = b, a

print(employee_name, employee_id, MAX_WEEKLY_HOURS)
print(city, country, q1, q2, q3)
print("after swap → a =", a, ", b =", b)

import keyword
print("\nReserved keywords you cannot use as names:\n", keyword.kwlist)
````

### 2.3 Dynamic typing
Python decides the type **at runtime from the value**, and the same name can later point to a value of a different type.

````python
# 🎯 Demo 2.3 — Dynamic typing
status = 200
print(status, type(status))

status = "OK"
print(status, type(status))

status = True
print(status, type(status))

# Python is dynamically BUT strongly typed: it will not silently mix types
try:
    print("Hours: " + 40)
except TypeError as e:
    print("TypeError →", e)
````

# Module 3 · Data Types & Type Conversion

### 3.1 Core built-in types
| Type | Example | Notes |
|---|---|---|
| `int` | `42`, `-7`, `1_000_000` | Whole numbers, unlimited size |
| `float` | `3.14`, `2.5e3` | Decimal numbers (IEEE-754, may have tiny rounding errors) |
| `str` | `"Pune"`, `'US'`, `"""multi"""` | Text, immutable |
| `bool` | `True`, `False` | Subclass of `int` (`True == 1`) |
| `NoneType` | `None` | "No value yet / nothing" |


````python
# 🎯 Demo 3.1 — The core types and type()
emp_count  = 350_000          # underscores improve readability
avg_rating = 4.6
office     = "Mumbai"
is_active  = True
manager    = None             # not assigned yet

for value in (emp_count, avg_rating, office, is_active, manager):
    print(f"{value!r:<10} → {type(value).__name__}")

print()
print("0.1 + 0.2 =", 0.1 + 0.2, " ← float rounding!")
print("round(0.1 + 0.2, 2) =", round(0.1 + 0.2, 2))
print("True + True =", True + True, " ← bool behaves like int")
````

### 3.3 Type checking & type conversion
- `type(x)` → returns the exact type
- `isinstance(x, (int, float))` → preferred for checks (supports several types)

| Conversion | Example | Result |
|---|---|---|
| `int()` | `int("42")`, `int(9.99)` | `42`, `9` (truncates, doesn't round) |
| `float()` | `float("3.5")`, `float(7)` | `3.5`, `7.0` |
| `str()` | `str(100)` | `"100"` |
| `bool()` | `bool(0)`, `bool("")`, `bool("no")` | `False`, `False`, **`True`** |

**Implicit conversion:** Python auto-promotes in mixed arithmetic → `3 + 2.0 == 5.0` (float).
**Falsy values:** `0`, `0.0`, `""`, `None`, `[]`, `{}`, `()` — everything else is truthy.


````python
# 🎯 Demo 3.4 — Explicit and implicit conversion
print(int("42"), int(9.99), int(-9.99))
print(float("3.5"), float(7))
print(str(100) + " employees")
print(3 + 2.0, type(3 + 2.0))                # implicit int → float

for v in [0, 1, "", "False", None, " "]:
    print(f"bool({v!r:<7}) = {bool(v)}")

# Conversion that fails
try:
    int("12.5")
except ValueError as e:
    print("\nValueError →", e, "→ use int(float('12.5')) instead:", int(float("12.5")))
````

# Module 4 · Operators & Expressions

| Category | Operators |
|---|---|
| Arithmetic | `+  -  *  /  //  %  **` |
| Comparison | `==  !=  >  <  >=  <=` (return `bool`) |
| Logical | `and  or  not` |
| Assignment | `=  +=  -=  *=  /=  //=  %=  **=` |
| Membership | `in  not in` |
| Identity | `is  is not` (same object in memory) |

**Precedence (high → low):** `()` → `**` → `* / // %` → `+ -` → comparisons → `not` → `and` → `or`


````python
# 🎯 Demo 4.1 — Arithmetic operators
a, b = 17, 5
print("a + b  =", a + b)
print("a - b  =", a - b)
print("a * b  =", a * b)
print("a / b  =", a / b, "  ← true division, always float")
print("a // b =", a // b, "    ← floor division")
print("a % b  =", a % b, "    ← remainder (modulus)")
print("a ** 2 =", a ** 2, "  ← power")

total_minutes = 487
print(f"\n{total_minutes} min = {total_minutes // 60} h {total_minutes % 60} min")
print("Precedence: 2 + 3 * 4 ** 2 =", 2 + 3 * 4 ** 2, "| (2 + 3) * 4 =", (2 + 3) * 4)
````

````python
# 🎯 Demo 4.2 — Comparison and logical operators
hours, rating = 42, 4.5

print(hours > 40, hours == 40, hours != 40)
print("Chained:", 35 <= hours <= 45)         # Pythonic range check

is_overtime  = hours > 40
is_top_rated = rating >= 4.5
print("Bonus eligible (and):", is_overtime and is_top_rated)
print("Any flag (or):       ", is_overtime or is_top_rated)
print("Not overtime (not):  ", not is_overtime)

# Short-circuit: the right side is skipped when the answer is already known
manager = None
print("Safe check:", manager is not None and manager.startswith("A"))
````

````python
# 🎯 Demo 4.3 — Assignment, membership and identity
balance = 1000
balance += 250    # 1250
balance -= 100    # 1150
balance *= 2      # 2300
balance //= 3     # 766
print("balance:", balance)

us_states = ["TX", "NY", "CA", "IL"]
print("'TX' in us_states:    ", "TX" in us_states)
print("'WA' not in us_states:", "WA" not in us_states)
print("'gem' in 'Capgemini': ", "gem" in "Capgemini")

x = [1, 2]; y = [1, 2]; z = x
print("x == y:", x == y, "| x is y:", x is y, "| x is z:", z is x)
````

# Module 5 · Conditional & Iterative Statements

### 5.1 `if` / `elif` / `else`
```python
if condition:
    ...
elif other_condition:
    ...
else:
    ...
```
Only the **first** true branch runs. Conditional (ternary) expression: `value = a if condition else b`.


````python
# 🎯 Demo 5.1 — Ticket priority routing
severity = 2
is_vip_client = True

if severity == 1:
    sla_hours = 1
elif severity == 2:
    sla_hours = 4
elif severity == 3:
    sla_hours = 24
else:
    sla_hours = 72

# Nested condition
if is_vip_client:
    if sla_hours > 2:
        sla_hours = sla_hours // 2

queue = "Priority" if sla_hours <= 4 else "Standard"   # ternary
print(f"Severity {severity} → SLA {sla_hours} h → {queue} queue")
````

### 5.2 Loops
- **`for`** — iterate over a sequence (list, string, `range()`, …). Use when you know *what* to loop over.
- **`while`** — repeat while a condition is true. Use when you don't know *how many* times.
- `range(start, stop, step)` — `stop` is **excluded**.
- `enumerate(seq)` gives index + value; `zip(a, b)` walks two lists together.


````python
# 🎯 Demo 5.2 — for loops
print(list(range(5)), list(range(2, 11, 2)), list(range(5, 0, -1)))

team = ["Asha", "Ben", "Carlos", "Divya"]
hours = [40, 38, 45, 42]

for name in team:
    print("Hello", name)

for i, name in enumerate(team, start=1):
    print(i, name)

for name, h in zip(team, hours):
    print(f"{name:<8}{h:>4} h")

total = 0
for h in hours:
    total += h
print("Team total:", total, "| average:", total / len(hours))
````

````python
# 🎯 Demo 5.3 — while loop: retry a deployment
import random
random.seed(7)

attempt = 0
max_attempts = 5
deployed = False

while not deployed and attempt < max_attempts:
    attempt += 1
    deployed = random.random() > 0.6       # simulate success/failure
    print(f"Attempt {attempt}: {'success ✅' if deployed else 'failed ❌'}")

print("Deployed!" if deployed else "Giving up after max attempts")
````

### 5.3 `break`, `continue`, `pass`
| Keyword | Effect |
|---|---|
| `break` | Exit the loop immediately |
| `continue` | Skip the rest of this iteration, go to the next |
| `pass` | Do nothing — a placeholder where syntax needs a statement |
| `for … else` | The `else` block runs only if the loop **did not** `break` |


````python
# 🎯 Demo 5.4 — break, continue, pass, for-else
tickets = ["INC101", "", "INC102", "SPAM", "INC103", "CRITICAL-INC104", "INC105"]

for t in tickets:
    if t == "":
        continue                 # skip blanks
    if t == "SPAM":
        pass                     # TODO: log spam later — placeholder for now
        continue
    if t.startswith("CRITICAL"):
        print("🚨 Escalating", t, "— stop processing queue")
        break
    print("Processing", t)

# for-else: search pattern
for emp_id in [101, 205, 309]:
    if emp_id == 999:
        print("Found!")
        break
else:
    print("Employee 999 not found in list")
````
