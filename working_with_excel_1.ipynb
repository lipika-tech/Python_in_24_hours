```python
# Reading the excel file
import openpyxl 

wb = openpyxl.load_workbook('videogamesales.xlsx')
```

```python
type(wb)
```

```python
# Now that the Excel file is loaded as a Python object, you need to tell the library which worksheet to access. 
 #There are two ways to do this:

# The first method is to simply call the active worksheet, which is the first sheet in the workbook, using the 
 # following line of code:
    
ws = wb.active    
```

```python
ws
```

```python
type(ws)
```

```python
# Alternatively, if you know the name of the worksheet, you can also access it by its name. 
 # I will be using the “vgsales” sheet in this section
    
ws = wb['vgsales']    
```

```python
type(ws)
```

```python
# Let’s now count the number of rows and columns in this worksheet
print('Total number of rows in our worksheet: '+str(ws.max_row)+'. And total number of columns in our worksheet: '+str(ws.max_column))
```

```python
# Reading the data from the cell
# To retrieve data from a specific cell with Openpyxl, you can type the cell’s value

print('The value in cell A1 is: ',ws['A1'].value)
```

```python
print('The value in cell A3 is: ',ws['A3'].value)
```

```python
# Reading the data from multiple cells

values = [ws.cell(row=1,column=i).value for i in range(1,ws.max_column+1)]
print(values)
```

```python
# Printing multiple rows in a column
# We will create a for loop to render the first ten rows in the “Name” column as a list. 

data=[ws.cell(row=i,column=2).value for i in range(2,12)]
print(data)
```

```python
# Printing first 10 rows in a range of columns

# reading data from a range of cells (from column 1 to 6)

# Creating an Empty List
my_list = list()


# Reads rows from row 1 to 11 (first 11 rows).
# Reads columns from 1 to 6 (first 6 columns).
# values_only=True ensures that only the cell values (not objects) are retrieved.
# Each row (a tuple of 6 values) is appended to my_list.
for value in ws.iter_rows(
    min_row=1, max_row=11, min_col=1, max_col=6, 
    values_only=True):
    my_list.append(value)


# Printing data into the format table
# Loops through the stored list my_list, unpacking 6 elements per row (ele1, ele2, ..., ele6).
# Uses .format() to align and space the printed output properly.

# {:<8} → Left-align first element (8 spaces wide).
# {:<35} → Left-align second element (35 spaces wide).
# And, so on.
for ele1,ele2,ele3,ele4,ele5,ele6 in my_list:
    (print ("{:<8}{:<35}{:<10}{:<10}{:<15}{:<15}".format(ele1,ele2,ele3,ele4,ele5,ele6)))
```

## Writing to an Excel file

```python
# There are two ways you can write to a file with Openpyxl.

# First, you can access the cell directly using its key:

ws['K1'] = 'Sum of Sales'
```

```python
# An alternative is to specify the row and column position of the cell that you’d like to write to:

ws.cell(row=1, column=11, value = 'Sum of Sales')
```

```python
# Every time you write to an Excel file with Openpyxl, you need to save your changes with the following 
 # line of code or they will not be reflected in the worksheet:

# Can give error if your workbook is open while saving data    
wb.save('videogamesales.xlsx')    
```

```python
# Creating new column

row_position = 2
col_position = 7


# Retrieves 4 consecutive values from row 2 in columns 7, 8, 9, and 10
# Adds them together to calculate total_sales
total_sales = ((ws.cell(row=row_position, column=col_position).value)+
               (ws.cell(row=row_position, column=col_position+1).value)+
               (ws.cell(row=row_position, column=col_position+2).value)+
               (ws.cell(row=row_position, column=col_position+3).value))

# Writes the calculated total_sales to row 2, column 11 in the Excel file
ws.cell(row=2,column=11).value=total_sales
wb.save('videogamesales.xlsx')
```

```python
# Notice that the total sales have been calculated in cell K2 for the first game in the worksheet
# You can check the sheet by opening it.
```

```python
# Similarly, let’s create a for loop to sum the sales values in every row:

row_position = 1

for i in range(1, ws.max_row):

    row_position += 1
    NA_Sales = ws.cell(row=row_position, column=7).value
    EU_Sales = ws.cell(row=row_position, column=8).value
    JP_Sales = ws.cell(row=row_position, column=9).value
    Other_Sales = ws.cell(row=row_position, column=10).value

    total_sales = (NA_Sales + EU_Sales + JP_Sales + Other_Sales)
    ws.cell(row=row_position, column=11).value = total_sales

wb.save("videogamesales.xlsx")
```

```python
# Appending new row
# To append a new row to the workbook, simply create a tuple with the values you’d like to include and write it to the sheet

new_row = (1,'The Legend of Zelda',1986,'Action','Nintendo',3.74,0.93,1.69,0.14,6.51,6.5)

ws.append(new_row)
    
wb.save('videogamesales.xlsx')
```

```python
# You can confirm that this data has been appended by printing the last row in the workbook:

values = [ws.cell(row=ws.max_row,column=i).value for i in range(1,ws.max_column+1)]
print(values)
```

```python
# Deleting the row

# ws.max_row Returns the last row number in the worksheet.
# Deletes 1 row, starting from the last row (ws.max_row).
ws.delete_rows(ws.max_row, 1) # row number, number of rows to delete

wb.save('videogamesales.xlsx')
```
